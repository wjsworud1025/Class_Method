## 0. 큰 그림: 왜 class를 배우나요?

### 🥮 비유 ① 붕어빵 틀과 붕어빵

- **class = 붕어빵 틀** (설계도, 아직 먹을 수 없음)
- **object = 붕어빵** (틀로 찍어낸 실제 결과물)
- 틀 하나로 붕어빵을 **여러 개** 찍을 수 있고, 붕어빵마다 속(팥/슈크림)은 **다르게** 넣을 수 있어요.

### 🏠 비유 ② 건축 설계도와 집

- 설계도(class)는 "방 3개, 화장실 2개"라는 **규칙**만 적은 종이입니다.
- 설계도로 지은 실제 집(object)은 **주소도, 벽지 색도 제각각**입니다.

### 🤖 AI 서비스와 연결

- 앞으로 쓸 `OpenAI` API 클라이언트, 챗봇, 데이터 처리기가 모두 **객체**입니다.
- "데이터 + 그 데이터를 다루는 기능"을 하나로 묶는 도구가 class입니다.

### 📌 핵심 용어

| 용어 | 뜻 | 비유 |
|---|---|---|
| class | 객체를 만드는 설계도 | 붕어빵 틀 |
| object (instance) | class로 만든 실제 결과물 | 붕어빵 |
| attribute | 객체가 가진 **데이터(특징)** | 붕어빵의 속, 크기 |
| method | 객체가 하는 **동작(함수)** | 붕어빵 굽기, 포장하기 |

## 1. class 선언하고 object 만들기

### 📖 개념

- `class 클래스이름:` 으로 선언합니다.
- 클래스 이름은 **단어 첫 글자를 대문자(파스칼)**로 씁니다. 예) `Dog`, `ChatBot`, `BankAccount`
- 안에 쓸 내용이 아직 없으면 `pass` (아무것도 안 함)를 적습니다.
- 객체는 **`클래스이름()`** 처럼 괄호를 붙여 만듭니다.

```python
# ✍️ 빈칸을 채우세요
class Dog:            # 빈칸: 클래스 이름 Dog
    pass               # 빈칸: 내용이 없을 때 쓰는 키워드

my_dog = Dog()
your_dog = Dog()        # 빈칸: 두 번째 객체 만들기

print(type(my_dog))
print(type(your_dog))
print(my_dog is your_dog)   # 결과를 예측해 본 뒤 실행!
```

### ✅ 체크포인트

- `Dog`는 **class**, `my_dog`는 **object** 입니다.
- 같은 틀로 만들어도 `my_dog`와 `your_dog`는 **서로 다른 객체**예요.

## 3. method 와 self: 객체의 동작 넣기

### 📖 개념

- 클래스 안에 만든 **함수**를 **method(메서드)** 라고 합니다.
- 메서드의 **첫 번째 매개변수는 항상 `self`** 입니다.
- `self` = **"지금 이 메서드를 부른 바로 그 객체 자신"**
- 객체 안의 속성을 쓸 때는 `self.속성이름` 으로 접근합니다.

### 🪞 비유: self 는 "나"

- 학생 A가 "내 이름은 철수야"라고 말할 때 "내"는 A를 가리키죠?
- `self.name` 은 "**나의** name" 입니다. 누가 호출하느냐에 따라 `self`가 달라집니다.

```python
# ✍️ 빈칸을 채우세요
class Dog:
    name = "이름없음"
    age = 0

    def bark(self):                           # 빈칸: 메서드의 첫 번째 매개변수
        print(f"{self.name}: 멍멍!")         # 빈칸: "나의" name 에 접근

    def birthday(self):
        self.age += 1                         # 빈칸: 내 나이 1 증가
        print(f"{self.name}의 나이가 {self.age}살이 되었어요 🎂")

my_dog = Dog()
my_dog.name = "홍"

my_dog.bark()
my_dog.birthday()
my_dog.birthday()
```

### 🔍 한 걸음 더: `self` 는 어디서 오나요?

우리가 쓰는 `my_dog.bark()` 는 파이썬 속에서 `Dog.bark(my_dog)` 로 바뀌어 실행됩니다.
즉, **`my_dog` 가 자동으로 `self` 에 전달**됩니다. 그래서 메서드를 정의할 때 `self` 를 꼭 적어야 해요.

```python
# ✍️ 같은 결과가 나오는 두 가지 호출 방식을 모두 실행해 보세요.
#    (1) 객체.메서드()
my_dog.bark()
#    (2) 클래스.메서드(객체)
Dog.bark(my_dog)
```

## 4. `__init__`: 객체를 만들면서 한 번에 세팅하기 (맛보기)

> 💡 (생성자와 소멸자)에서 자세히 배웁니다. 지금은 "이런 게 있구나" 정도만!

- 지금까지는 객체를 만든 **뒤에** 이름과 나이를 따로 넣었어요. 번거롭죠.
- `__init__` 은 객체가 만들어지는 **그 순간 자동 실행**되는 특별한 메서드(생성자)입니다.
- 붕어빵을 구울 때 **반죽에 팥을 바로 넣는 단계**라고 생각하세요.

```python
# ✍️ 빈칸을 채우세요
class Dog:
    def __init__(self, name, age):        # 빈칸: 생성자 이름 (밑줄 2개씩!)
        self.name = "초코"              # 빈칸: 받은 name 저장
        self.age = 3               # 빈칸: 받은 age 저장

    def bark(self):
        print(f"{self.name}: 멍멍!")

my_pet = Dog("홍", 2)
print(my_dog.name, my_dog.age)
my_dog.bark()
```

## 6. 자주 하는 실수 TOP 5

| 증상 (에러 메시지) | 원인 | 해결 |
|---|---|---|
| `TypeError: bark() takes 0 positional arguments but 1 was given` | 메서드에 `self` 를 안 적음 | `def bark(self):` |
| `NameError: name 'name' is not defined` | 메서드 안에서 `self.` 없이 속성 사용 | `self.name` 으로 접근 |
| `AttributeError: 'Dog' object has no attribute 'name'` | 속성을 만든 적이 없음 / 오타 | 속성 이름 확인, `__init__` 에서 설정 |
| `TypeError: __init__() missing 1 required positional argument` | 객체 만들 때 값을 덜 넣음 | `Dog("초코", 3)` 처럼 모두 전달 |
| `my_dog.bark` 로 썼는데 아무 일도 안 일어남 | 괄호 `()` 누락 | `my_dog.bark()` |

> 💬 **에러 메시지는 힌트입니다.** 마지막 줄을 소리 내서 읽어 보는 습관을 들이세요!

```python
# ✍️ 에러 체험
# 1) self 없이 hello() 메서드를 가진 Broken 클래스를 만들고
class Broken():
    def hello(self):
        print("hello~")
# 2) Broken().hello() 를 try / except TypeError as e 로 실행해 에러 메시지를 확인
try:
    Broken().hello()
except TypeError as e:
    print("에러메시지 ->", e)
# 3) 에러 메시지의 어떤 단어가 "self 를 빼먹었다"는 힌트인지 찾아보기
```

## 7. 🛠️ 단계별 실습 과제

### 실습 1 나만의 학생 클래스

**요구사항**

- `Student` 클래스를 만든다. (`__init__` 으로 `name`, `score` 받기)
- `introduce()` 메서드: `"안녕하세요, 저는 OOO이고 점수는 N점입니다."` 출력
- `is_pass()` 메서드: 점수가 60점 이상이면 `True`, 아니면 `False` 리턴
- 학생 객체를 2개 만들어 확인한다.

```python
# ✍️ 실습 1: 빈칸을 채워 완성하세요
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def introduce(self):
        print(f"안녕하세요, 저는 {self.name}이고 점수는 {self.score}점 입니다.")

    def is_pass(self):
        return self.score >=60

a=Student("전재경", 50)
a.introduce()
a.is_pass()
```

### 실습 2  통장 클래스

**요구사항**

- `BankAccount` 클래스 (`owner`, `balance=0` 을 `__init__` 으로 받기)
- `deposit(amount)`: 입금 → 잔액 증가
- `withdraw(amount)`: 출금 → 잔액이 부족하면 `"잔액이 부족합니다"` 출력하고 출금하지 않기
- `show()`: `"OOO님의 잔액: N원"` 출력

```python
# ✍️ 실습 2: BankAccount 클래스를 완성하세요
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if self.balance-amount < 0:
            print("잔액이 부족합니다.")
        else:
            self.balance -= amount

    def show(self):
        print(f"{self.owner}님의 잔액 :{self.balance}원")

a = BankAccount("하늘", 10000)
a.deposit(20000)
a.show()
a.withdraw(40000)
a.show()
# 기대 결과: 잔액이 부족합니다 → 하늘님의 잔액: 12000원
```

### 실습 3 미니 챗봇 클래스 🤖



**요구사항**

- `ChatBot` 클래스 (`__init__` 에서 `name` 을 받고, 대화 기록용 빈 리스트 `history` 만들기)
- `say(user_message)`: `"[봇이름] 기억할게요 → '메시지'"` 를 출력하고, `history` 에 `(사용자 메시지, 봇 답변)` 저장
- `show_history()`: 지금까지의 대화 기록을 번호와 함께 출력
- 챗봇 객체 2개를 만들어 **기록이 서로 섞이지 않는지** 확인

```python
# ✍️ 실습 3: ChatBot 클래스를 직접 작성하세요 (골격만 제공)
class ChatBot:
    def __init__(self, name, history=[]):
        self.name = name
        self.history = history

    def say(self, user_message):
        print(f"{self.name} 기억할게요 -> {user_message}")
        self.history.append(user_message)

    def show_history(self):
       for i, h in enumerate(self.history):
           print(f"{i+1}번 메시지: {h}")

# 테스트 코드를 직접 작성하세요
# 1) bot_a = ChatBot("코코"),  bot_b = ChatBot("모모")
bot_a = ChatBot("교주")
bot_b = ChatBot("두두")
# 2) bot_a.say 2번, bot_b.say 1번
bot_a.say("두두 안녕!")
bot_a.say("두두 오늘은 뭐할까?")
bot_b.say("두두 교규, 조아!")
# 3) 두 봇의 show_history() 호출 → 기록이 섞이지 않았는지 확인
bot_a.show_history()
```
