## 🟢 문제 1 (기본) 직사각형 클래스 `Rectangle` — *실습 1 유사*

**요구사항**

- 생성자 `__init__(self, width, height, name="사각형")`
  - 가로·세로가 **음수면 0으로 바로잡아** 저장 (생성자 안에서 검사!)
- `get_area()` : 넓이(가로 × 세로) **리턴**
- `get_perimeter()` : 둘레(2 × (가로 + 세로)) **리턴**
- `get_name()` : 이름 **리턴**
- 소멸자 : `"이름 is deleted"` 출력

**예상 출력**
```
50 30 사각형
0 2 잘못된사각형
사각형 is deleted
잘못된사각형 is deleted
```

````python
# ✍️ 문제 1: Rectangle 클래스를 완성하세요
class Rectangle:
    def __init__(self, width, height, name="사각형"):
        # TODO: 음수 검사 후 width, height, name 저장
        if width < 0:
            width = 0
        if height < 0:
            height = 0

        self.width = width
        self.height = height
        self.name = name

    def get_area(self):
        return self.width * self.height

    def get_perimeter(self):
        return 2*(self.width + self.height)

    def get_name(self):
        return self.name

    def __del__(self):
        print(f"{self.name} is deleted")

r1 = Rectangle(10, 5)
r2 = Rectangle(-3, 1, name="잘못된사각형")
print(r1.get_area(), r1.get_perimeter(), r1.get_name())
print(r2.get_area(), r2.get_perimeter(), r2.get_name())
del r1
del r2
````

````python
import io, contextlib
def _run(f):
    """f() 실행 중 print 된 내용을 문자열로 돌려주는 채점 도우미"""
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        f()
    return buf.getvalue()

# 🧪 자동 채점 (수정하지 마세요)
_r = Rectangle(4, 3)
assert _r.get_name() == "사각형", "name 기본값 확인"
assert _r.get_area() == 12 and _r.get_perimeter() == 14
_n = Rectangle(-1, -2, "N")
assert (_n.width, _n.height) == (0, 0), "음수는 0으로!"
def _f():
    global _r
    del _r
assert _run(_f).strip() == "사각형 is deleted", "소멸자 출력 확인"
_k = Rectangle(height=2, width=7, name="K")
assert _k.get_area() == 14, "이름으로 값 넣기도 동작해야 해요"
print("🎉 문제 1 통과!")
````

## 🟢 문제 2 (기본) DB 연결 흉내 내기 `DBConnection` — *실습 2 유사*

**요구사항**

- 생성자 `__init__(self, db_name, user="guest")`
  - `db_name`, `user`, `query_count = 0` 저장
  - `"[DB명] 사용자 접속"` 출력
- `query(self, sql)`
  - `query_count` 1 증가
  - `"[DB명] 'SQL' 실행 완료 (N번째)"` 문자열을 **리턴**
- 소멸자 : `"[DB명] user 접속 해제 (쿼리 N회)"` 출력

**예상 출력**
```
[shop] admin 접속
[shop] 'SELECT * FROM item' 실행 완료 (1번째)
[shop] 'DELETE FROM cart' 실행 완료 (2번째)
[shop] admin 접속 해제 (쿼리 2회)
[test] guest 접속
[test] guest 접속 해제 (쿼리 0회)
```

````python
# ✍️ 문제 2: DBConnection 클래스를 완성하세요
class DBConnection:
    def __init__(self, db_name, user="guest"):
        # TODO 1: db_name, user, query_count = 0 저장
        self.db_name = db_name
        self.user = user
        self.query_count = 0
        # TODO 2: 접속 메시지 출력
        print(f"[{self.db_name}] {self.user} 접속")
    def query(self, sql):
        # TODO 3: query_count 1 증가
        self.query_count += 1
        # TODO 4: 문자열 return
        return f"[{self.db_name}] '{sql}' 실행 완료 ({self.query_count}번째)"

    def __del__(self):
        # TODO 5: 접속 해제 메시지 출력
        print(f"[{self.db_name}] {self.user} 접속 해제 (쿼리 {self.query_count}회)")

conn = DBConnection("shop", user="admin")
print(conn.query("SELECT * FROM item"))
print(conn.query("DELETE FROM cart"))
del conn
guest = DBConnection("test")
del guest
````

````python
import io, contextlib
def _run(f):
    """f() 실행 중 print 된 내용을 문자열로 돌려주는 채점 도우미"""
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        f()
    return buf.getvalue()

# 🧪 자동 채점 (수정하지 마세요)
_box = []
assert _run(lambda: _box.append(DBConnection("t"))).strip() == "[t] guest 접속", "생성 메시지 확인"
_c = _box.pop()
assert _c.query_count == 0
assert _c.query("A") == "[t] 'A' 실행 완료 (1번째)", "query 는 print 가 아니라 return!"
_c.query("B"); _c.query("C")
def _f():
    global _c
    del _c
assert _run(_f).strip() == "[t] guest 접속 해제 (쿼리 3회)"
print("🎉 문제 2 통과!")
````

## 🟢 문제 3 (기본) 게임 캐릭터 `GameCharacter` — *실습 3 유사*

**요구사항**

- 생성자 `__init__(self, name, job="전사", hp=100)`
  - `name`, `job`, `hp` 저장, 가방 `inventory = []` 생성
  - 생성되자마자 `"[이름] 직업(HP N) 등장!"` 출력
- `pick(self, item)` : `inventory` 에 추가하고 `"[이름] 아이템 획득!"` 출력
- `__del__` : `"[이름] 퇴장. 아이템 N개: [목록]"` 출력
- 캐릭터 2개(기본값 사용 1개, `job`·`hp` 직접 지정 1개)를 만들어 아이템을 줍고 삭제

**예상 출력**
```
[아서] 전사(HP 100) 등장!
[멀린] 마법사(HP 70) 등장!
[아서] 검 획득!
[아서] 방패 획득!
[멀린] 지팡이 획득!
[아서] 퇴장. 아이템 2개: ['검', '방패']
[멀린] 퇴장. 아이템 1개: ['지팡이']
```

````python
# ✍️ 문제 3: GameCharacter 클래스를 완성하세요
class GameCharacter:
    def __init__(self, name, job="전사", hp=100):
        self.name = name
        self.job = job
        self.hp = hp
        self.inventory = []
        print(f"[{self.name}] {self.job}(HP {self.hp}) 등장!")

    def pick(self, item):
        self.inventory.append(item)
        print(f"[{self.name}] {item} 획득!")

    def __del__(self):
        print(f"[{self.name}] 퇴장. 아이템 {len(self.inventory)}개: {self.inventory}")

# 테스트 코드를 직접 작성하세요 (예상 출력과 같게)
arthur = GameCharacter("아서")
merlin = GameCharacter("멀린", job="마법사", hp=70)

arthur.pick("검")
arthur.pick("방패")
merlin.pick("지팡이")

del arthur
del merlin
````

````python
import io, contextlib
def _run(f):
    """f() 실행 중 print 된 내용을 문자열로 돌려주는 채점 도우미"""
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        f()
    return buf.getvalue()

# 🧪 자동 채점 (수정하지 마세요)
_box = []
assert _run(lambda: _box.append(GameCharacter("A"))).strip() == "[A] 전사(HP 100) 등장!"
_a = _box.pop()
_b = GameCharacter("B", hp=50, job="궁수")
assert (_b.job, _b.hp) == ("궁수", 50)
_run(lambda: _a.pick("x"))
assert _a.inventory == ["x"] and _b.inventory == [], "가방이 섞였어요! inventory 는 생성자에서"
def _f():
    global _a
    del _a
assert _run(_f).strip() == "[A] 퇴장. 아이템 1개: ['x']"
print("🎉 문제 3 통과!")
````

## 🔴 심화 1 (난이도 상) 접속자 수 세기 `Session` — 생성자·소멸자 + 클래스 변수

웹 서비스의 **현재 접속자 수**를 생성자(로그인)·소멸자(로그아웃)로 자동 관리합니다.

**요구사항**

- 클래스 변수 `active = 0` (현재 접속 중인 세션 수), `total_created = 0` (지금까지 만들어진 세션 수)
- 생성자 `__init__(self, user)`
  - `user` 가 빈 문자열이면 `ValueError("사용자 이름이 비어 있습니다")` 를 **발생**(raise)
  - 정상이면 `total_created`, `active` 를 1씩 증가, 세션 번호 `sid = total_created` 저장
  - `"[로그인] #번호 이름 (접속 중: N)"` 출력
- 소멸자 `__del__`
  - `active` 1 감소, `"[로그아웃] #번호 이름 (접속 중: N)"` 출력

> ⚠️ **함정**: `__init__` 에서 에러가 나도 **이미 만들어진 빈 객체의 `__del__` 은 호출됩니다!**
> 실패한 세션 때문에 `active` 가 줄어들거나 `AttributeError` 가 나지 않도록 막아야 합니다.
> (힌트: 생성자 **맨 첫 줄**에 "정상 생성 여부" 표시 속성을 두세요.)

**(1) 출력 예측** — 아래 시나리오의 출력을 **먼저 예측**해 적고, 구현 후 실행해 비교하세요.

```python
s1 = Session("민수")
s2 = Session("지아")
alias = s1             # 같은 객체에 이름표 2개
del s1                 # ①
s2 = Session("하늘")   # ②

def visit():
    guest = Session("손님")
    print("둘러보는 중...")
visit()                # ③

try:
    Session("")        # ④
except ValueError as e:
    print("에러:", e)

print("현재 접속:", Session.active, "/ 누적:", Session.total_created)

```

**(1) 출력 예측** 📌 ①~⑤ 각각에서 로그아웃 메시지가 나올까요? 누구의?



````python
# ✍️ 심화 1: Session 클래스를 작성하세요
class Session:
    active = 0
    total_created = 0

    def __init__(self, user):
        self.created = False
        if user == "":
            raise ValueError("사용자 이름이 비어 있습니다")

        self.user = user
        Session.active += 1
        Session.total_created += 1
        self.sid=self.total_created

        self.created = True
        print(f"[로그인] #{self.sid} {self.user} (접속 중:{Session.active})")

    def __del__(self):
        if self.created:
            Session.active -= 1
            print(f"[로그아웃] #{self.sid} {self.user} (접속 중: {Session.active})")

Session.active = 0; Session.total_created = 0     # (재실행 대비 초기화)
s1 = Session("민수")
s2 = Session("지아")
alias = s1
del s1
s2 = Session("하늘")

def visit():
    guest = Session("손님")
    print("둘러보는 중...")
visit()

try:
    Session("")
except ValueError as e:
    print("에러:", e)

print("현재 접속:", Session.active, "/ 누적:", Session.total_created)
del alias
````

````python
import io, contextlib
def _run(f):
    """f() 실행 중 print 된 내용을 문자열로 돌려주는 채점 도우미"""
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        f()
    return buf.getvalue()

# 🧪 자동 채점 (수정하지 마세요)
Session.active = 0; Session.total_created = 0
_box = []
_out = _run(lambda: _box.append(Session("A")))
assert _out.strip() == "[로그인] #1 A (접속 중: 1)", _out
_s1 = _box.pop(); _s2 = _s1
def _f1():
    global _s1
    del _s1
assert _run(_f1) == "", "이름표가 남아 있으면 소멸자가 실행되면 안 돼요"
def _bad():
    try:
        Session("")
    except ValueError:
        pass
_out = _run(_bad)
assert Session.active == 1 and Session.total_created == 1, f"실패한 세션이 카운트에 영향을 줬어요 (active={Session.active})"
assert "로그아웃" not in _out, "실패한 세션은 로그아웃 메시지가 없어야 해요"
def _f2():
    global _s2
    del _s2
assert _run(_f2).strip() == "[로그아웃] #1 A (접속 중: 0)"
_run(lambda: (lambda: Session("tmp"))())      # 함수 안 임시 객체 → 바로 소멸
assert Session.active == 0 and Session.total_created == 2
print("🎉 심화 1 통과!")
````

## 🔴 심화 2 (난이도 상) 파일 로거 `FileLogger` — 진짜 자원 정리하기

소멸자의 실제 쓰임새인 **파일 닫기**를 구현합니다. 단, "소멸자에만 의존하지 않고 **직접 닫을 수도 있는**" 안전한 구조로 만듭니다.

**요구사항**

- 생성자 `__init__(self, path)`
  - `path` 저장, 줄 수 `lines = 0`, 닫힘 여부 `closed = False`
  - `open(path, "w", encoding="utf-8")` 로 파일을 열어 `self.file` 에 저장
  - `"[로그 시작] path"` 출력
- `log(self, msg, level="INFO")`
  - 이미 닫혔으면 `"이미 닫힌 로거입니다"` 출력 후 `False` 리턴
  - 아니면 `"001 [INFO] 메시지"` 형식(번호 3자리)으로 한 줄 쓰고 `True` 리턴
- `close(self)`
  - 이미 닫혔으면 **아무 것도 하지 않음** (두 번 호출해도 안전)
  - 파일을 닫고 `closed = True`, `"[로그 종료] path (N줄)"` 출력
- `__del__` : `close()` 를 호출 → **직접 닫지 않았을 때만** 자동으로 닫힘 메시지가 나와야 함
- 🌟 **보너스**: `__enter__` / `__exit__` 를 추가해 `with FileLogger(...) as lg:` 로 쓸 수 있게 만들기

**예상 출력**
```
[로그 시작] app.log
[로그 종료] app.log (3줄)
이미 닫힌 로거입니다
--- del 실행 ---
--- 파일 내용 ---
001 [INFO] 서버 시작
002 [WARN] 메모리 80%
003 [ERROR] 연결 실패
[로그 시작] auto.log
[로그 종료] auto.log (1줄)
```

````python
# ✍️ 심화 2: FileLogger 클래스를 작성하세요
class FileLogger:
    def __init__(self, path):
        self.path = path
        self.lines = 0
        self.closed = False
        self.file = open(path, "w", encoding="utf-8")

        print(f"[로그 시작] {path}")

    def log(self, msg, level="INFO"):
        if self.closed:
            print("이미 닫힌 로거입니다")
            return False

        self.lines += 1
        self.file.write(f"{self.lines:03d} [{level}] {msg}\n")
        return True

    def close(self):
        if self.closed:
            return
        self.file.close()
        self.closed=True
        print(f"[로그 종료] {self.path} ({self.lines}줄)")

    def __del__(self):
        if hasattr(self, "file"):
            self.close()

    # 🌟 보너스: __enter__, __exit__
    def __enter__(self):
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        self.close()
        return False

lg = FileLogger("app.log")
lg.log("서버 시작")
lg.log("메모리 80%", level="WARN")
lg.log("연결 실패", "ERROR")
lg.close()
lg.log("닫은 뒤 기록")          # 닫힌 로거
print("--- del 실행 ---")
del lg                          # 이미 닫았으므로 메시지 없음

print("--- 파일 내용 ---")
print(open("app.log", encoding="utf-8").read(), end="")

auto = FileLogger("auto.log")
auto.log("close 를 깜빡함")
del auto                        # 소멸자가 대신 닫아 줌
````

````python
import io, contextlib
def _run(f):
    """f() 실행 중 print 된 내용을 문자열로 돌려주는 채점 도우미"""
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        f()
    return buf.getvalue()

# 🧪 자동 채점 (수정하지 마세요)
import os
_box = []
assert _run(lambda: _box.append(FileLogger("_t.log"))).strip() == "[로그 시작] _t.log"
_l = _box.pop()
assert _l.log("a") is True and _l.log("b", "DEBUG") is True and _l.lines == 2
assert _run(_l.close).strip() == "[로그 종료] _t.log (2줄)"
assert _run(_l.close) == "", "두 번째 close 는 아무 출력도 없어야 해요"
assert _run(lambda: _l.log("x")).strip() == "이미 닫힌 로거입니다" and _l.lines == 2
def _f1():
    global _l
    del _l
assert _run(_f1) == "", "이미 닫았으면 소멸자에서 또 메시지가 나오면 안 돼요"
assert open("_t.log", encoding="utf-8").read() == "001 [INFO] a\n002 [DEBUG] b\n", "파일 형식 확인"
_m = FileLogger("_t2.log"); _m.log("z")
def _f2():
    global _m
    del _m
assert _run(_f2).strip() == "[로그 종료] _t2.log (1줄)", "close 안 했으면 소멸자가 닫아야 해요"
assert open("_t2.log", encoding="utf-8").read() == "001 [INFO] z\n"
os.remove("_t.log"); os.remove("_t2.log")
print("🎉 심화 2 통과!")

# 🌟 보너스 채점 (with 문)
if hasattr(FileLogger, "__enter__"):
    def _w():
        with FileLogger("_t3.log") as w:
            w.log("in with")
        print("블록 밖")
    assert _run(_w).splitlines()[-2:] == ["[로그 종료] _t3.log (1줄)", "블록 밖"], "with 블록이 끝나는 순간 닫혀야 해요"
    os.remove("_t3.log")
    print("🌟 보너스 통과!")
````
