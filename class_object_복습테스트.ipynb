## 🟢 문제 1 상품 클래스

**요구사항**

- `Product` 클래스를 만든다. `__init__` 으로 `name`, `price`, `stock`(기본값 `0`) 받기
- `info()` : `"[상품] 이름 | 가격: N원 | 재고: N개"` 출력
- `is_sold_out()` : 재고가 0이면 `True`, 아니면 `False` **리턴**
- `discount_price(rate)` : `rate`% 할인된 가격을 **정수(int)로 리턴** (예: 10000원, 15% → 8500)
- 상품 객체 2개를 만들어 확인한다.

**예상 출력**
```
[상품] 키보드 | 가격: 30000원 | 재고: 5개
[상품] 마우스 | 가격: 15000원 | 재고: 0개
False True
25500
```

````python
# ✍️ 문제 1: Product 클래스를 완성하세요
class Product:
    def __init__(self, name, price, stock=0):
        # TODO: 속성 3개 저장
        self.name = name
        self.price = price
        self.stock = stock

    def info(self):
        # TODO
        print(f"[상품] {self.name} | 가격: {self.price}원 | 재고: {self.stock}개")

    def is_sold_out(self):
        # TODO: True / False 리턴
        if self.stock == 0:
            return True
        else:
            return False

    def discount_price(self, rate):
        # TODO: 할인된 가격을 int 로 리턴
        return int(self.price * (1 - rate / 100))


p1 = Product("키보드", 30000, 5)
p2 = Product("마우스", 15000)
p1.info()
p2.info()
print(p1.is_sold_out(), p2.is_sold_out())
print(p1.discount_price(15))
````

````python
# 🧪 자동 채점 (수정하지 마세요)
_a = Product("A", 10000, 3); _b = Product("B", 2000)
assert (_a.name, _a.price, _a.stock) == ("A", 10000, 3)
assert _b.stock == 0, "stock 기본값은 0"
assert _a.is_sold_out() is False and _b.is_sold_out() is True, "is_sold_out 은 print 가 아니라 return!"
assert _a.discount_price(15) == 8500 and isinstance(_a.discount_price(15), int)
assert _a.price == 10000, "discount_price 는 원래 price 를 바꾸면 안 돼요"
print("🎉 문제 1 통과!")
````

## 🟢 문제 2 (기본) 포인트 카드 클래스 

**요구사항**

- `PointCard` 클래스 (`owner`, `point=0` 을 `__init__` 으로 받기)
- `earn(pay)` : 결제금액의 **5%를 정수로** 적립 → `"N점 적립"` 출력
- `use(amount)` : 포인트가 부족하면 `"포인트가 부족합니다"` 출력하고 **차감하지 않기**, 충분하면 차감 후 `"N점 사용"` 출력
- `show()` : `"OOO님의 포인트: N점"` 출력

**예상 출력**
```
500점 적립
150점 적립
포인트가 부족합니다
600점 사용
하늘님의 포인트: 50점
```

````python
# ✍️ 문제 2: PointCard 클래스를 완성하세요
class PointCard:
    # TODO: __init__, earn, use, show
    def __init__(self, card, point=0):
        self.card = card
        self.point = point

    def earn(self, pay):
        earned = int(pay * 5 / 100)
        self.point += earned
        print(f"{earned}점 적립")

    def use(self, amount):
        if amount > self.point:
            print("포인트가 부족합니다.")
        else:
            self.point -= amount
            print(f"{amount}점 사용")

    def show(self):
        print(f"{self.card}님의 포인트 : {self.point}점")

card = PointCard("하늘")
card.earn(10000)
card.earn(3000)
card.use(1000)
card.use(600)
card.show()
````

````python
# 🧪 자동 채점 (수정하지 마세요)
_c = PointCard("테스트")
assert _c.point == 0
_c.earn(10000); _c.earn(999)          # 500 + 49
assert _c.point == 549, f"5% 정수 적립 확인 (현재 {_c.point})"
_c.use(10000)
assert _c.point == 549, "부족하면 차감하지 않아야 해요"
_c.use(549)
assert _c.point == 0
_d = PointCard("다른사람", 100)
assert _d.point == 100 and _c.point == 0, "객체끼리 포인트가 섞이면 안 돼요"
print("🎉 문제 2 통과!")
````

## 🟢 문제 3 할 일 목록 클래스 

**요구사항**

- `TodoList` 클래스 (`__init__` 에서 `owner` 를 받고, 할 일 저장용 빈 리스트 `tasks` 만들기)
- `add(task)` : `tasks` 에 `[할일, False]` 형태로 저장하고 `"[OOO] 추가: 할일"` 출력
- `complete(number)` : **1번부터 시작하는 번호**의 할 일을 완료(True) 처리. 없는 번호면 `"없는 번호입니다"` 출력
- `show()` : 완료는 `[✔]`, 미완료는 `[ ]` 로 번호와 함께 출력
- `count_left()` : 남은(미완료) 할 일 개수 **리턴**
- 객체 2개를 만들어 **목록이 서로 섞이지 않는지** 확인

**예상 출력**
```
[민수] 추가: 장보기
[민수] 추가: 숙제
[지아] 추가: 운동
없는 번호입니다
--- 민수의 할 일 ---
1. [✔] 장보기
2. [ ] 숙제
--- 지아의 할 일 ---
1. [ ] 운동
남은 할 일: 1 1
```

````python
# ✍️ 문제 3: TodoList 클래스를 완성하세요
class TodoList:
    def __init__(self, owner):
        self.owner = owner
        self.tasks = []

    def add(self, task):
        self.tasks.append([task, False])
        print(f"[{self.owner}] 추가: {task}")

    def complete(self, number):
        if number < 1 or number > len(self.tasks):
            print("없는 번호입니다")
            return

        self.tasks[number - 1][1] = True

    def show(self):
        print(f"--- {self.owner}의 할 일 ---")
        for number, (task, done) in enumerate(self.tasks, start=1):
            mark = "✔" if done else " "
            print(f"{number}. [{mark}] {task}")

    def count_left(self):
        return sum(1 for task, done in self.tasks if not done)


a = TodoList("민수")
b = TodoList("지아")
a.add("장보기")
a.add("숙제")
b.add("운동")
a.complete(1)
b.complete(5)
a.show()
b.show()
print("남은 할 일:", a.count_left(), b.count_left())
````

````python
# 🧪 자동 채점 (수정하지 마세요)
_x = TodoList("X"); _y = TodoList("Y")
_x.add("a"); _x.add("b"); _y.add("c")
assert len(_x.tasks) == 2 and len(_y.tasks) == 1, "목록이 섞였어요! tasks 를 어디서 만들었나요?"
_x.complete(2)
assert _x.tasks[1][1] is True and _x.tasks[0][1] is False, "번호는 1부터 시작"
_x.complete(0); _x.complete(3)        # 없는 번호 → 에러 없이 메시지만
assert _x.count_left() == 1 and _y.count_left() == 1
print("🎉 문제 3 통과!")
````

## 🔴 심화 1 기록이 섞이는 챗봇 고치기 + 클래스 변수

아래 코드는 봇마다 대화 기록을 따로 갖게 하려고 했지만 **기록이 섞이는 버그**가 있습니다.

**(1) 원인 설명** — 아래 마크다운 셀에 한두 문장으로 쓰세요.

**(2) 새 `ChatBot` 클래스 작성 요구사항**

- **클래스 변수** `bot_count` : 지금까지 만들어진 봇 수 (객체 생성 시 1 증가)
- **클래스 변수** `total_messages` : **모든 봇**이 받은 메시지 총합
- `__init__(name, rules, max_history=3)`
  - `rules` : `{"키워드": "답변"}` 딕셔너리 (봇마다 다를 수 있음)
  - `history` : 인스턴스마다 독립된 리스트
- `say(msg)` : `rules` 의 키워드가 `msg` 안에 **포함**되면 해당 답변, 없으면 `"잘 모르겠어요."`
  - `"[봇이름] 답변"` 출력, `(msg, 답변)` 을 history 에 저장
  - history 가 `max_history` 를 넘으면 **가장 오래된 기록부터 삭제** (최근 N개만 유지)
  - 답변 문자열을 **리턴**
- `summary()` : `{"name": 이름, "count": 현재 history 개수, "last": 마지막 사용자 메시지 또는 None}` **리턴**

> ⚠️ 함정 주의: `self.total_messages += 1` 은 클래스 변수를 바꿀까요?

**예상 출력**
```
[코코] 반가워요!
[코코] 오늘은 맑아요.
[코코] 잘 모르겠어요.
[코코] 반가워요!
[모모] 냠냠, 밥 먹자!
{'name': '코코', 'count': 3, 'last': '안녕 또 왔어'}
{'name': '모모', 'count': 1, 'last': '배고파'}
봇 수: 2 / 전체 메시지: 5
```

**(1) 원인 설명** 📌 내 답:



````python
# ✍️ 심화 1: ChatBot 클래스를 작성하세요
class ChatBot:
    bot_count = 0
    total_messages = 0

    def __init__(self, name, rules, max_history=3):
        self.name = name
        self.rules = rules
        self.max_history = max_history
        self.history = []

        ChatBot.bot_count +=1

    def say(self, msg):
        answer = "잘 모르겠어요."
        for k in self.rules:
            if k in msg:
                answer=self.rules[k]
                break

        print(f"[{self.name}] {answer}")
        self.history.append((msg, answer))
        ChatBot.total_messages += 1

        if len(self.history) > self.max_history:
            self.history.pop(0)

        return answer

    def summary(self):
        last = None
        if len(self.history) > 0:
            last_record = self.history
            last_record = self.history[-1]
            last = last_record[0]

        return{
            "name" : self.name,
            "count" : len(self.history),
            "last" : last,
        }


ChatBot.bot_count = 0; ChatBot.total_messages = 0   # (재실행 대비 초기화)
coco = ChatBot("코코", {"안녕": "반가워요!", "날씨": "오늘은 맑아요."})
momo = ChatBot("모모", {"배고파": "냠냠, 밥 먹자!"}, max_history=5)

coco.say("안녕?")
coco.say("내일 날씨 어때?")
coco.say("파이썬 알아?")
coco.say("안녕 또 왔어")
momo.say("배고파")
print(coco.summary())
print(momo.summary())
print("봇 수:", ChatBot.bot_count, "/ 전체 메시지:", ChatBot.total_messages)
````

````python
# 🧪 자동 채점 (수정하지 마세요)
ChatBot.bot_count = 0; ChatBot.total_messages = 0
_p = ChatBot("P", {"hi": "hello"}, max_history=2)
_q = ChatBot("Q", {})
assert ChatBot.bot_count == 2, "bot_count 는 클래스 변수로 증가해야 해요"
assert _p.history is not _q.history, "history 가 공유되고 있어요!"
assert _p.say("oh hi there") == "hello"
assert _q.say("hi") == "잘 모르겠어요.", "rules 도 봇마다 따로!"
_p.say("1"); _p.say("2")
assert [m for m, _ in _p.history] == ["1", "2"], "최근 max_history 개만 남겨야 해요"
assert ChatBot.total_messages == 4, f"모든 봇 메시지 합계 (현재 {ChatBot.total_messages})"
assert "total_messages" not in vars(_p), "self.total_messages += 1 함정에 빠졌어요!"
assert _q.summary() == {"name": "Q", "count": 1, "last": "hi"}
assert ChatBot("R", {}).summary()["last"] is None
print("🎉 심화 1 통과!")
````

## 🔴 심화 2 은행 시스템 — 객체끼리 협력하기

지금까지는 객체 혼자 일했습니다. 이번엔 **객체가 다른 객체를 다루는** 구조를 만듭니다.

**(A) `Account` 클래스**

- `__init__(owner, balance=0)` + 거래내역 리스트 `log`
- `deposit(amount)` / `withdraw(amount)` : 성공하면 `True`, 실패하면 `False` **리턴**
  - 금액이 0 이하 → `"금액이 올바르지 않습니다"` 출력, 실패
  - 출금 시 잔액 부족 → `"잔액이 부족합니다"` 출력, 실패
  - 성공 시 `log` 에 `"입금 +5000"` / `"출금 -3000"` 형태로 기록
- `transfer(target, amount)` : **다른 `Account` 객체** `target` 에게 이체
  - 실패 조건은 출금과 동일 (실패 시 **양쪽 잔액 모두 변화 없음**)
  - 성공 시 내 log `"이체 -2000 → 철수"`, 상대 log `"이체 +2000 ← 영희"`

**(B) `Bank` 클래스** — 여러 `Account` 를 관리

- `__init__(name)` : `accounts` 딕셔너리 `{owner: Account}`
- `open_account(owner, balance=0)` : 새 계좌를 만들어 등록하고 **리턴**. 이미 있으면 `"이미 존재하는 계좌입니다"` 출력, `None` 리턴
- `find(owner)` : 계좌 객체 또는 `None` 리턴
- `transfer(from_owner, to_owner, amount)` : 둘 중 하나라도 없으면 `"계좌를 찾을 수 없습니다"` 출력, `False`. 있으면 `Account.transfer` 에 **위임**
- `total_balance()` : 은행 전체 잔액 합계 리턴
- `richest()` : 잔액이 가장 많은 고객 이름 리턴

**예상 출력**
```
이미 존재하는 계좌입니다
잔액이 부족합니다
계좌를 찾을 수 없습니다
금액이 올바르지 않습니다
영희: 13000원 ['입금 +5000', '이체 -2000 → 철수']
철수: 5000원 ['이체 +2000 ← 영희', '출금 -500']
총액: 18000 / 최고 부자: 영희
```

````python
# ✍️ 심화 2: Account, Bank 클래스를 작성하세요
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.log = []

    def deposit(self, amount):
        if amount <= 0:
            return False
        self.balance += amount
        self.log.append(f"입금 +{amount}")
        return True

    def withdraw(self, amount):
        if amount <= 0:
            print("출금액은 0보다 커야 합니다.")
            return False
        if amount > self.balance:
            print("잔액이 부족합니다.")
            return False
        self.balance -= amount
        self.log.append(f"출금 - {amount}")
        return True

    def transfer(self, target, amount):
        if amount <= 0:
            print("출금액은 0보다 커야 합니다.")
            return False
        if amount > self.balance:
            print("잔액이 부족합니다.")
            return False
        self.balance -= amount
        target.balance +=amount

        self.log.append(f"이체 -{amount} → {target.owner}")
        target.log.append(f"이체 +{amount} ← {self.owner}")
        return True

class Bank:
    def __init__(self, name):
        self.name = name
        self.accounts = {}

    def open_account(self, owner, balance=0):
        if owner in self.accounts:
            print("이미 존재하는 계좌입니다.")
            return None
        account = Account(owner, balance)
        self.accounts[owner] = account
        return account

    def find(self, owner):
        if owner in self.accounts:
            return self.accounts[owner]
        return None

    def transfer(self, from_owner, to_owner, amount):
        send = self.find(from_owner)
        receive = self.find(to_owner)

        if send is None or receive is None:
            print("계좌를 찾을 수 없습니다")
            return False
        return send.transfer(receive, amount)

    def total_balance(self):
        total = 0

        for owner in self.accounts:
            account = self.accounts[owner]
            total += account.balance
        return total

    def richest(self):
        richest_name = None
        max_balance = -1
        
        for owner in self.accounts:
            balance = self.accounts[owner].balance
            if balance > max_balance:
                max_balance = balance
                richest_name=owner
        return richest_name

bank = Bank("파이썬은행")
bank.open_account("영희", 10000)
bank.open_account("철수", 3500)
bank.open_account("영희")                 # 중복
bank.find("영희").deposit(5000)
bank.transfer("철수", "영희", 99999)       # 잔액 부족
bank.transfer("영희", "민지", 1000)        # 없는 계좌
bank.find("철수").withdraw(-100)          # 잘못된 금액
bank.transfer("영희", "철수", 2000)
bank.find("철수").withdraw(500)

for name in ["영희", "철수"]:
    acc = bank.find(name)
    print(f"{acc.owner}: {acc.balance}원 {acc.log}")
print(f"총액: {bank.total_balance()} / 최고 부자: {bank.richest()}")
````
