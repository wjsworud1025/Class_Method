## 1. 생성자 `__init__`: 태어나는 순간 자동 실행!

### 📖 개념

- `__init__` 은 객체가 **만들어지는 바로 그 순간 자동으로 실행**되는 특별한 메서드입니다.
- 이름 앞뒤에 밑줄(`_`)이 **2개씩** 필요합니다. (`_init_` ❌ → `__init__` ✅)
- 첫 번째 매개변수는 **항상 `self`**, 나머지는 객체를 만들 때 넘겨줄 값입니다.

### 🥮 비유

- 붕어빵을 **굽는 순간 반죽에 팥을 넣는 단계**
- 학생이 **입학하는 순간 학생증이 발급**되는 것과 같아요. 입학(객체 생성)과 동시에 이름·학번(속성)이 정해집니다.

### 🔄 실행 순서

```
Dog("초코", 3)  →  ① 빈 객체 생성  →  ② __init__(self, "초코", 3) 자동 호출  →  ③ my_dog 에 저장
```

````python
class Dog:
    def __init__(self, name, age):
        print("init() 가 자동으로 실행됐어요!")
        self.name = name
        self.age = age

my_dog = Dog("초코", 3)
print(my_dog.name, my_dog.age)
````

### 💡 `name` 과 `self.name` 은 다른 거예요

| 표기 | 정체 | 비유 |
|---|---|---|
| `name` | `__init__` 안에서만 쓰는 **임시 입력값** (매개변수) | 접수창구에 낸 신청서 |
| `self.name` | 객체가 **계속 보관**하는 속성 | 서류함에 보관된 내 서류 |

→ 신청서(`name`)를 받아서 **서류함(`self.name`)에 넣어야** 나중에도 꺼내 쓸 수 있습니다.

### 📖 기본값 매개변수 & 이름으로 값 넣기

- 매개변수에 `= 값` 을 붙이면 **안 넣었을 때 쓸 기본값**이 됩니다.
- 객체를 만들 때 `name="콩이"` 처럼 **이름을 붙여서** 넣으면 순서가 바뀌어도 됩니다.

````python
# ✍️ 위 설명대로 직접 만들어 보세요
# 1) __init__ 에 name 기본값 "이름없음", age 기본값 0 을 주기
class Dog:
    def __init__(self, name="이름없음", age=0):
        self.name = name
        self.age = age
# 2) 객체 3개 만들기: 값 없이 / name 만 / age=2, name="콩이" (순서 바꿔서)
dog1 = Dog()
dog2 = Dog("콩이")
dog3 = Dog(age=2, name="콩이")
# 3) 3개 객체의 name, age 를 각각 print
print(dog1.name, dog1.age, )
print(dog2.name, dog2.age, )
print(dog3.name, dog3.age, )
````

### ⚠️ 에러 체험: 값을 덜 넣으면?

기본값이 없는 매개변수에 값을 안 넣으면 파이썬이 알려 줍니다. **에러 메시지 읽는 연습**을 해 봅시다.

````python
# ✍️ 위 Cat 클래스를 선언하고, age 를 빠뜨린 채 객체를 만들어
class Cat:
    def __init__(self, name, age):
        self.name = name
        if age < 0:
            age =0
        self.age = age
#    에러 메시지가 무엇인지 확인해 보세요. (try / except TypeError as e 사용)
try:
    nabi = Cat("나비", -1)
    print(nabi.name, nabi.age)
except TypeError as t:
    print("에러메세지 : ", t )
#    힌트: 에러 메시지에 'missing 1 required positional argument' 가 보이면 성공!
````

### 🧠 생성자 안에서는 "검사"도 할 수 있어요

값이 이상하면 기본값으로 바로잡을 수도 있습니다. (예: 나이가 음수면 0으로)

````python
# ✍️ class Dog 는 생성자와 소멸자 두 기능이 오버라이딩 되어있다. name을 전달받아 이름을 출력합니다.
#    소멸자는 소멸대상의 참조값을 전달받아 소멸시킵니다.
class Dog:
    def __init__(self, name, age):
        print("init() 시작")
        self.name = name
        if age < 0:
            age =0
        self.age = age
        print(f"{self.name} 태어났어요")
        print("init() 끝")
    def __del__(self):
        print("del() 시작")
        print(f"{self.name}이 사라졌어요")
        print("del() 끝")

dog1 = Dog("초코", 2)
print("----del 전")
del dog1
print("----del 후")
````

## 2. 소멸자 `__del__`: 사라지는 순간 자동 실행!

### 📖 개념

- `__del__` 은 객체가 **메모리에서 사라질 때 자동으로 실행**되는 메서드입니다.
- 주로 **뒷정리**(연결 끊기, 파일 닫기, 로그 남기기)를 할 때 씁니다.

### 💡 비유

- **퇴실할 때 불 끄고 문 잠그기** 🔌
- **붕어빵을 다 먹은 뒤 포장지 버리기** 🗑️

### 🔑 핵심 규칙

객체는 "**아무 변수도 그 객체를 가리키지 않게 되면**" 사라집니다.
(변수 = 객체에 붙은 이름표라고 생각하세요. 이름표가 0개가 되면 객체는 정리됩니다.)

### 📌 `__del__` 이 실행되는 3가지 상황

| 상황 | 예시 |

| ③ 함수가 끝나서 지역 변수가 사라짐 / 프로그램 종료 | 함수 안에서 만든 객체 |

## 3. 🛠️ 단계별 실습 과제

### 실습 1  정삼각형 클래스 `Figure`

**요구사항**

- attribute: 한 변의 길이 `length`, 이름 `name`
- 생성자에서 attribute 값 설정
- `get_area()`: 정삼각형 넓이 리턴 → 공식 **(√3 / 4) × 한 변²**
- `get_name()`: 이름 리턴
- 소멸자: 객체가 사라질 때 `object is deleted` 출력
- 길이 10 → 넓이 약 **43.30** 이 나오는지 검산하기

````python
# ✍️ 실습 1: 빈칸을 채워 완성하세요
import math

class Figure:
    def __init__(self, length, name):                  # 빈칸: 변의 길이, 이름
        self.length = length
        self.name = name

    def get_area(self):
        return (math.sqrt(3) / 4) * self.length ** 2   # 빈칸: 공식의 분모

    def get_name(self):
        return self.name

    def __del__(self):                                  # 빈칸: 소멸자 이름
        print("object is deleted")
triangle = Figure(10, "triangle")
print(triangle.get_area(),triangle.get_name())
del triangle
````

### 실습 2  AI API 클라이언트 흉내 내기 🤖



**요구사항**

- 클래스 이름: `ApiClient`
- 생성자 `__init__(self, service_name)`
  - `service_name`, `call_count = 0` 저장
  - `"[서비스명] 연결 시작"` 출력
- `ask(self, prompt)` 메서드
  - `call_count` 1 증가
  - `"[서비스명] 답변: '프롬프트' 에 대한 가짜 응답"` 문자열을 **리턴**
- 소멸자 `__del__`: `"[서비스명] 연결 종료 (총 N회 호출)"` 출력

````python
# ✍️ 실습 2: 요구사항을 보고 ApiClient 클래스를 직접 완성하세요
class ApiClient:
    def __init__(self, service_name, call_count = 0):
        # TODO 1: service_name 저장, call_count = 0 만들기
        self.service_name = service_name
        self.call_count = call_count
        # TODO 2: "[서비스명] 연결 시작" 출력
        print(f"{self.service_name} 연결 시작")

    def ask(self, prompt):
        # TODO 3: call_count 1 증가
        self.call_count += 1
        # TODO 4: 가짜 응답 문자열 return
        return f"[{self.service_name}] 답변: {prompt}에 대한 가짜 응답"

    def __del__(self):
        # TODO 5: "[서비스명] 연결 종료 (총 N회 호출)" 출력
        print(f"[{self.service_name}] 연결 종료 (총 {self.call_count}회 호출)")

# 아래는 테스트 코드입니다. (수정 금지)
client = ApiClient("FakeGPT")
print(client.ask("안녕?"))
print(client.ask("파이썬이 뭐야?"))
del client
# 기대 결과 마지막 줄: [FakeGPT] 연결 종료 (총 2회 호출)
````

### 실습 3  챗봇 v2 만들기

> `ChatBot` 에 **생성자 기본값**과 **소멸자 정리 기능**을 추가합니다.

**요구사항**

- `ChatBot.__init__(self, name, greeting="안녕하세요!")`
  - `name`, `greeting` 저장, 대화 기록 `history = []` 생성
  - 생성되자마자 `"[이름] 인사말"` 출력
- `say(self, user_message)`
  - `"[이름] 기억할게요 → '메시지'"` 출력
  - `history` 에 메시지 저장 (`append`)
- `__del__`: `"[이름] 안녕히 계세요. 오늘 대화는 N개였어요."` 출력
- 챗봇 2개(하나는 인사말 기본값 사용, 하나는 직접 지정)를 만들어 대화 후 삭제해 보기

````python
# ✍️ 실습 3: 요구사항을 보고 ChatBot 클래스를 직접 작성하세요
# (골격만 제공합니다. pass 자리를 채워 보세요)

class ChatBot:
    def __init__(self, name, greeting="안녕하세요!"):
#       TODO: 속성 3개(name, greeting, history) 만들고 인사말 출력
        self.name = name
        self.greeting = greeting
        self.history = []
        print(f"[{self.name}] {self.greeting}")
    def say(self, user_message):
#       TODO: 출력 + history 에 저장
        print(f"[{self.name}] {user_message}")
        self.history.append(user_message)
    def __del__(self):
#       TODO: 대화 개수 포함한 작별 인사 출력
        print(f"[{self.name}] 전체 대화 개수: {len(self.history)}개")
        print("안녕히 가세요 대화는 삭제됩니다.")

bot_a = ChatBot("코코")
bot_b = ChatBot("모모", greeting="반가워요!")

bot_a.say("천국과 지옥이 뭔가요")
bot_a.say("코코는 천국에서 기다릴게요")
bot_b.say("모모는 나쁜 강아지니까 지옥에 가나요?")

del bot_a
del bot_b
# 테스트 코드를 직접 작성하세요
# 1) bot_a = ChatBot("코코")  /  bot_b = ChatBot("모모", greeting="반가워요!")
# 2) bot_a.say 2번, bot_b.say 1번
# 3) del bot_a, del bot_b
````
